```python
from glob import glob
import os
from meshtosdf import mesh_to_sdf_npz
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn
import torch.nn.functional as F
import plotly.graph_objects as go
from skimage import measure
```

```python
def convert_all_meshes(mesh_dir="meshes", out_dir="sdfs_npz",
                       resolution=128, padding=5):
    os.makedirs(out_dir, exist_ok=True)

    mesh_paths = []
    for ext in ("obj", "ply", "stl"):
        mesh_paths += glob(os.path.join(mesh_dir, f"*.{ext}"))

    assert len(mesh_paths) > 0, "No meshes found."

    sdf_paths = []
    for mesh_path in mesh_paths:
        name = os.path.splitext(os.path.basename(mesh_path))[0]
        out_path = os.path.join(out_dir, f"{name}_sdf_{resolution}.npz")
        mesh_to_sdf_npz(mesh_path, out_path,
                        resolution=resolution, padding=padding)
        sdf_paths.append(out_path)

    return sdf_paths
```

```python
def pad_crop_to_shape_3d(vol: np.ndarray, target_shape=(128,128,128), pad_value=None):
    """
    Center pad/crop a 3D volume to target shape.

    Args:
      vol: (D,H,W)
      target_shape: (Dt,Ht,Wt)
      pad_value: value used for padding. If None, use a large positive value from vol.

    Returns:
      out: (Dt,Ht,Wt)
    """
    assert vol.ndim == 3
    D, H, W = vol.shape
    Dt, Ht, Wt = target_shape

    if pad_value is None:
        # outside SDF should be positive; use a robust large positive value
        pad_value = float(np.percentile(vol, 95))

    out = np.full(target_shape, pad_value, dtype=vol.dtype)

    # Compute start indices in source and destination (centered)
    def compute_slices(s, t):
        if s >= t:
            src0 = (s - t) // 2
            dst0 = 0
            length = t
        else:
            src0 = 0
            dst0 = (t - s) // 2
            length = s
        return src0, dst0, length

    sd0, dd0, ld = compute_slices(D, Dt)
    sh0, dh0, lh = compute_slices(H, Ht)
    sw0, dw0, lw = compute_slices(W, Wt)

    out[dd0:dd0+ld, dh0:dh0+lh, dw0:dw0+lw] = vol[sd0:sd0+ld, sh0:sh0+lh, sw0:sw0+lw]
    return out
```

```python
def create_slice_observation(sdf, n_slices_z=4, thickness=1, margin=3, seed=None):
    rng = np.random.default_rng(seed)
    D = sdf.shape[0]

    partial = np.zeros_like(sdf, dtype=np.float32)
    mask = np.zeros_like(sdf, dtype=np.float32)

    zs = rng.choice(
        np.arange(margin, D - margin),
        size=n_slices_z,
        replace=False
    )

    for z in zs:
        z0 = max(0, z - thickness // 2)
        z1 = min(D, z0 + thickness)
        partial[:, :, z0:z1] = sdf[:, :, z0:z1]
        mask[:, :, z0:z1] = 1.0

    return partial, mask, zs.tolist()
```

```python
class SDFSliceDataset(Dataset):
    def __init__(self, npz_paths, target_shape=(128,128,128),
                 n_slices_z=4, thickness=1, seed=0):
        self.npz_paths = npz_paths
        self.target_shape = target_shape
        self.n_slices_z = n_slices_z
        self.thickness = thickness
        self.rng = np.random.default_rng(seed)

        self.sdfs = []
        self.names = []

        for p in npz_paths:
            data = np.load(p)
            sdf = data["sdf"].astype(np.float32)

            # Force fixed shape
            sdf_fixed = pad_crop_to_shape_3d(sdf, target_shape=target_shape)

            self.sdfs.append(sdf_fixed)
            self.names.append(os.path.basename(p))

    def __len__(self):
        return len(self.sdfs)

    def __getitem__(self, idx):
        sdf = self.sdfs[idx]
        name = self.names[idx]

        seed = int(self.rng.integers(0, 1_000_000))
        partial, mask, zs = create_slice_observation(
            sdf,
            n_slices_z=self.n_slices_z,
            thickness=self.thickness,
            margin=3,
            seed=seed
        )

        x = np.stack([partial, mask], axis=0).astype(np.float32)
        y = sdf[None].astype(np.float32)

        meta = {"name": name, "z_slices": zs}
        return torch.from_numpy(x), torch.from_numpy(y), meta
```

```python
def visualize_dataset_sample_3d(dataset, idx=0, level=0.0, max_points=20000):
    x, y, meta = dataset[idx]

    partial = x[0].numpy()
    mask = x[1].numpy()
    sdf = y[0].numpy()

    # GT surface
    verts, faces, _, _ = measure.marching_cubes(sdf, level=level)
    vx, vy, vz = verts.T
    i, j, k = faces.T

    # Observed voxels (from slices)
    xs, ys, zs = np.where(mask > 0.5)
    pts = np.stack([xs, ys, zs], axis=1)

    if len(pts) > max_points:
        sel = np.random.choice(len(pts), size=max_points, replace=False)
        pts = pts[sel]

    fig = go.Figure()

    fig.add_trace(go.Mesh3d(
        x=vx, y=vy, z=vz,
        i=i, j=j, k=k,
        opacity=0.35,
        name="GT surface (SDF=0)"
    ))

    fig.add_trace(go.Scatter3d(
        x=pts[:,0], y=pts[:,1], z=pts[:,2],
        mode="markers",
        marker=dict(size=2, color="red", opacity=0.6),
        name="Observed slice voxels"
    ))

    fig.update_layout(
        title=f"Dataset sample {idx} | {meta['name']} | slices={meta['z_slices']}",
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=40, b=0)
    )
    fig.show()
```

```python
def visualize_gt_with_slice_planes(dataset, idx=0, level=0.0, plane_opacity=0.15):
    x, y, meta = dataset[idx]
    sdf = y[0].numpy()
    D = sdf.shape[0]

    # GT surface
    verts, faces, _, _ = measure.marching_cubes(sdf, level=level)
    fig = go.Figure()
    fig.add_trace(go.Mesh3d(
        x=verts[:,0], y=verts[:,1], z=verts[:,2],
        i=faces[:,0], j=faces[:,1], k=faces[:,2],
        opacity=0.45, name="GT surface"
    ))

    # Slice planes (here using Z slices; adapt if you also do X/Y)
    zs = meta.get("z_slices", [])
    for z0 in zs:
        # plane grid
        xx, yy = np.meshgrid(np.arange(D), np.arange(D), indexing="ij")
        zz = np.full_like(xx, z0)

        fig.add_trace(go.Surface(
            x=xx, y=yy, z=zz,
            opacity=plane_opacity,
            showscale=False,
            name=f"z={z0}",
        ))

    fig.update_layout(
        title=f"GT + slice planes | {meta['name']} | z_slices={zs}",
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=40, b=0)
    )
    fig.show()
```

```python
class DoubleConv3D(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv3d(in_ch, out_ch, 3, padding=1), nn.ReLU(inplace=True),
            nn.Conv3d(out_ch, out_ch, 3, padding=1), nn.ReLU(inplace=True),
        )
    def forward(self, x):
        return self.net(x)

def match_size(src, ref):
    """
    Make src spatial size match ref (center crop or pad).
    src, ref: (B,C,D,H,W)
    """
    sd, sh, sw = src.shape[-3:]
    rd, rh, rw = ref.shape[-3:]

    # Crop if too big
    d0 = max((sd - rd) // 2, 0)
    h0 = max((sh - rh) // 2, 0)
    w0 = max((sw - rw) // 2, 0)
    src = src[:, :, d0:d0+rd, h0:h0+rh, w0:w0+rw]

    # Pad if too small
    sd, sh, sw = src.shape[-3:]
    pd = max(rd - sd, 0)
    ph = max(rh - sh, 0)
    pw = max(rw - sw, 0)

    if pd > 0 or ph > 0 or pw > 0:
        # pad format for 5D: (W_left, W_right, H_left, H_right, D_left, D_right)
        src = F.pad(src, (pw//2, pw - pw//2, ph//2, ph - ph//2, pd//2, pd - pd//2))

    return src

class UNet3D(nn.Module):
    def __init__(self, in_ch=2, out_ch=1, base=16):
        super().__init__()
        self.down1 = DoubleConv3D(in_ch, base)
        self.pool1 = nn.MaxPool3d(2)
        self.down2 = DoubleConv3D(base, base*2)
        self.pool2 = nn.MaxPool3d(2)
        self.mid   = DoubleConv3D(base*2, base*4)

        self.up2   = nn.ConvTranspose3d(base*4, base*2, 2, stride=2)
        self.dec2  = DoubleConv3D(base*4, base*2)
        self.up1   = nn.ConvTranspose3d(base*2, base, 2, stride=2)
        self.dec1  = DoubleConv3D(base*2, base)

        self.out   = nn.Conv3d(base, out_ch, 1)

    def forward(self, x):
        x1 = self.down1(x)              # (B,base, D,H,W)
        x2 = self.down2(self.pool1(x1)) # (B,2base,...)
        xm = self.mid(self.pool2(x2))   # (B,4base,...)

        x  = self.up2(xm)
        x  = match_size(x, x2)
        x  = self.dec2(torch.cat([x, x2], dim=1))

        x  = self.up1(x)
        x  = match_size(x, x1)
        x  = self.dec1(torch.cat([x, x1], dim=1))

        return self.out(x)
```

```python
def gradient_3d_forward(f):
    """
    f: (B,1,D,H,W)
    returns dx,dy,dz with forward differences
    """
    dx = f[:, :, 1:, :, :] - f[:, :, :-1, :, :]
    dy = f[:, :, :, 1:, :] - f[:, :, :, :-1, :]
    dz = f[:, :, :, :, 1:] - f[:, :, :, :, :-1]
    return dx, dy, dz


def eikonal_loss_3d(f):
    """
    Enforces ||∇f|| ≈ 1 with consistent tensor shapes.
    """
    dx, dy, dz = gradient_3d_forward(f)

    # Crop to common shape: (B,1,D-1,H-1,W-1)
    dx_c = dx[:, :, :, :-1, :-1]
    dy_c = dy[:, :, :-1, :, :-1]
    dz_c = dz[:, :, :-1, :-1, :]

    grad_mag = torch.sqrt(dx_c**2 + dy_c**2 + dz_c**2 + 1e-6)
    return torch.mean(torch.abs(grad_mag - 1.0))

def laplacian_loss_3d(f):
    """
    Discrete 3D Laplacian squared (6-neighborhood).
    f: (B,1,D,H,W)
    """
    lap = (
        -6.0 * f
        + torch.roll(f,  1, dims=2) + torch.roll(f, -1, dims=2)
        + torch.roll(f,  1, dims=3) + torch.roll(f, -1, dims=3)
        + torch.roll(f,  1, dims=4) + torch.roll(f, -1, dims=4)
    )
    return torch.mean(lap**2)

def curvature_loss_3d(f):
    """
    Penalize second derivatives (curvature proxy).
    f: (B,1,D,H,W)
    """
    # second derivative along W (x-like)
    dxx = f[:, :, :, :, :-2] - 2*f[:, :, :, :, 1:-1] + f[:, :, :, :, 2:]
    # second derivative along H (y-like)
    dyy = f[:, :, :, :-2, :] - 2*f[:, :, :, 1:-1, :] + f[:, :, :, 2:, :]
    # second derivative along D (z-like)
    dzz = f[:, :, :-2, :, :] - 2*f[:, :, 1:-1, :, :] + f[:, :, 2:, :, :]

    return torch.mean(dxx**2) + torch.mean(dyy**2) + torch.mean(dzz**2)
```

```python
def loss_function_3d(
    pred, gt, partial, mask,
    λ_rec=1.0,
    λ_eik=0.1,
    λ_mask=50.0,
    λ_smooth=0.02,
    λ_curv=0.02,
):
    """
    pred, gt, partial, mask: (B,1,D,H,W)
    """
    # Reconstruction
    L_rec = torch.mean(torch.abs(pred - gt))

    # Eikonal
    L_eik = eikonal_loss_3d(pred)

    # Consistency where observed (slices)
    L_mask = torch.mean(((pred - partial) ** 2) * mask)

    # Smoothness
    L_smooth = laplacian_loss_3d(pred)

    # Curvature (2nd derivatives)
    L_curv = curvature_loss_3d(pred)

    return (
        λ_rec * L_rec
        + λ_eik * L_eik
        + λ_mask * L_mask
        + λ_smooth * L_smooth
        + λ_curv * L_curv
    )
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

dataset = SDFSliceDataset(
    npz_paths=sdf_paths,
    target_shape=(128,128,128),
    n_slices_z=4,
    thickness=1,
    seed=0
)

x, gt, meta = dataset[0]
print("x:", x.shape, "gt:", gt.shape, meta["name"])

```

```python
from torch.utils.data.dataloader import default_collate

def collate_keep_meta(batch):
    xs, ys, metas = zip(*batch)
    return default_collate(xs), default_collate(ys), list(metas)
```

```python
loader = DataLoader(dataset, batch_size=2, shuffle=True, collate_fn=collate_keep_meta)

model = UNet3D().to(device)
opt = torch.optim.Adam(model.parameters(), lr=1e-3)

for epoch in range(5):
    model.train()
    total = 0.0
    for x, gt, meta in loader:
        x, gt = x.to(device), gt.to(device)
        pred = model(x)

        partial = x[:, 0:1]
        mask    = x[:, 1:2]

        loss = loss_function_3d(
            pred, gt, partial, mask,
            λ_rec=1.0,
            λ_eik=0.1,
            λ_mask=50.0,
            λ_smooth=0.02,
            λ_curv=0.02
        )

        opt.zero_grad()
        loss.backward()
        opt.step()

        total += loss.item()

    print(f"Epoch {epoch+1} | loss = {total/len(loader):.4f}")
```

```python
def visualize_prediction_vs_gt(
    model,
    dataset,
    idx=0,
    device="cpu",
    level=0.0,
    shift=160,
):
    """
    Visualize GT vs Prediction SDF surfaces side-by-side.

    Args:
      model: trained UNet3D
      dataset: SDFSliceDatasetFixedSize
      idx: index of the sample
      device: cpu / cuda
      level: SDF iso-level (usually 0)
      shift: translation applied to prediction for side-by-side view
    """
    model.eval()
    with torch.no_grad():
        x, gt, meta = dataset[idx]
        x = x.unsqueeze(0).to(device)   # (1,2,D,H,W)
        pred = model(x).cpu()[0, 0].numpy()
        gt = gt[0].numpy()

    # Marching cubes
    v_gt, f_gt, _, _ = measure.marching_cubes(gt, level=level)
    v_pr, f_pr, _, _ = measure.marching_cubes(pred, level=level)

    # Shift prediction to the right
    v_pr_shifted = v_pr.copy()
    v_pr_shifted[:, 0] += shift

    fig = go.Figure()

    # Ground truth
    fig.add_trace(go.Mesh3d(
        x=v_gt[:, 0], y=v_gt[:, 1], z=v_gt[:, 2],
        i=f_gt[:, 0], j=f_gt[:, 1], k=f_gt[:, 2],
        opacity=0.5,
        name="Ground Truth"
    ))

    # Prediction
    fig.add_trace(go.Mesh3d(
        x=v_pr_shifted[:, 0], y=v_pr_shifted[:, 1], z=v_pr_shifted[:, 2],
        i=f_pr[:, 0], j=f_pr[:, 1], k=f_pr[:, 2],
        opacity=0.5,
        name="Prediction"
    ))

    fig.update_layout(
        title=f"GT vs Prediction | {meta['name']} | slices={meta['z_slices']}",
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=40, b=0)
    )

    fig.show()
```

```python
def safe_marching_cubes(vol, preferred_level=0.0):
    vmin, vmax = float(vol.min()), float(vol.max())
    if vmin <= preferred_level <= vmax:
        level = preferred_level
    else:
        # choose a level inside the range
        level = 0.5 * (vmin + vmax)
    verts, faces, _, _ = measure.marching_cubes(vol, level=level)
    return verts, faces, level, (vmin, vmax)

def visualize_prediction_vs_gt_safe(model, dataset, idx=0, device="cpu", shift=160):
    model.eval()
    with torch.no_grad():
        x, gt, meta = dataset[idx]
        pred = model(x.unsqueeze(0).to(device)).cpu()[0,0].numpy()
        gt_np = gt[0].numpy()

    v_gt, f_gt, lvl_gt, rg_gt = safe_marching_cubes(gt_np, preferred_level=0.0)
    v_pr, f_pr, lvl_pr, rg_pr = safe_marching_cubes(pred, preferred_level=0.0)

    v_pr_shifted = v_pr.copy()
    v_pr_shifted[:, 0] += shift

    fig = go.Figure()
    fig.add_trace(go.Mesh3d(
        x=v_gt[:,0], y=v_gt[:,1], z=v_gt[:,2],
        i=f_gt[:,0], j=f_gt[:,1], k=f_gt[:,2],
        opacity=0.5,
        name=f"GT (level={lvl_gt:.3f})"
    ))
    fig.add_trace(go.Mesh3d(
        x=v_pr_shifted[:,0], y=v_pr_shifted[:,1], z=v_pr_shifted[:,2],
        i=f_pr[:,0], j=f_pr[:,1], k=f_pr[:,2],
        opacity=0.5,
        name=f"Pred (level={lvl_pr:.3f})"
    ))

    fig.update_layout(
        title=(
            f"{meta['name']} | slices={meta['z_slices']}<br>"
            f"GT range={rg_gt} | Pred range={rg_pr} | Pred level used={lvl_pr:.3f}"
        ),
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=60, b=0)
    )
    fig.show()
```
