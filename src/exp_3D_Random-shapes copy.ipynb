```python
def convert_all_meshes_to_sdf(
    mesh_dir="meshes",
    out_dir="sdfs_npz",
    resolution=128,
    padding=5,
    exts=("obj", "ply", "stl")
):
    """
    Convert all mesh files in mesh_dir to SDF .npz files using mesh_to_sdf_npz.

    Returns:
      npz_paths: list of output npz paths
    """
    os.makedirs(out_dir, exist_ok=True)

    mesh_paths = []
    for ext in exts:
        mesh_paths += glob(os.path.join(mesh_dir, f"*.{ext}"))

    if len(mesh_paths) == 0:
        raise FileNotFoundError(f"No mesh files found in '{mesh_dir}' with extensions {exts}")

    npz_paths = []
    for mesh_path in mesh_paths:
        name = os.path.splitext(os.path.basename(mesh_path))[0]
        out_path = os.path.join(out_dir, f"{name}_sdf_{resolution}.npz")
        mesh_to_sdf_npz(mesh_path, out_path, resolution=resolution, padding=padding)
        npz_paths.append(out_path)

    return npz_paths
```

```python
def pad_crop_to_shape_3d(vol: np.ndarray, target_shape=(128,128,128), pad_value=None):
    """
    Center pad/crop a 3D volume to target shape.
    """
    assert vol.ndim == 3
    D, H, W = vol.shape
    Dt, Ht, Wt = target_shape

    if pad_value is None:
        pad_value = float(np.percentile(vol, 95))  # positive outside

    out = np.full(target_shape, pad_value, dtype=vol.dtype)

    def compute_slices(s, t):
        if s >= t:
            src0 = (s - t) // 2
            dst0 = 0
            length = t
        else:
            src0 = 0
            dst0 = (t - s) // 2
            length = s
        return src0, dst0, length

    sd0, dd0, ld = compute_slices(D, Dt)
    sh0, dh0, lh = compute_slices(H, Ht)
    sw0, dw0, lw = compute_slices(W, Wt)

    out[dd0:dd0+ld, dh0:dh0+lh, dw0:dw0+lw] = vol[sd0:sd0+ld, sh0:sh0+lh, sw0:sw0+lw]
    return out
```

```python
def create_axis_aligned_loop_observation(
    sdf: np.ndarray,
    axis: str = "z",
    n_slices: int = 30,
    loop_band: float = 1.0,
    margin: int = 5,
):
    D, H, W = sdf.shape
    partial = np.zeros_like(sdf, dtype=np.float32)
    mask = np.zeros_like(sdf, dtype=np.float32)

    if axis == "z":
        idxs = np.linspace(margin, W - 1 - margin, n_slices).astype(int)
        for z in idxs:
            loop = (np.abs(sdf[:, :, z]) <= loop_band)
            partial[:, :, z][loop] = sdf[:, :, z][loop]
            mask[:, :, z][loop] = 1.0
        info = {"axis": "z", "slice_indices": idxs.tolist()}

    elif axis == "y":
        idxs = np.linspace(margin, H - 1 - margin, n_slices).astype(int)
        for y in idxs:
            loop = (np.abs(sdf[:, y, :]) <= loop_band)
            partial[:, y, :][loop] = sdf[:, y, :][loop]
            mask[:, y, :][loop] = 1.0
        info = {"axis": "y", "slice_indices": idxs.tolist()}

    elif axis == "x":
        idxs = np.linspace(margin, D - 1 - margin, n_slices).astype(int)
        for x in idxs:
            loop = (np.abs(sdf[x, :, :]) <= loop_band)
            partial[x, :, :][loop] = sdf[x, :, :][loop]
            mask[x, :, :][loop] = 1.0
        info = {"axis": "x", "slice_indices": idxs.tolist()}

    else:
        raise ValueError("axis must be 'x', 'y', or 'z'")

    return partial, mask, info
```

```python
def extract_loops_as_polylines(
    sdf: np.ndarray,
    axis: str = "z",
    n_slices: int = 30,
    level: float = 0.0,
    margin: int = 5,
    simplify_stride: int = 2
):
    D, H, W = sdf.shape
    loops = []

    if axis == "z":
        idxs = np.linspace(margin, W - 1 - margin, n_slices).astype(int)
        for z in idxs:
            img = sdf[:, :, z]
            contours = measure.find_contours(img, level=level)
            for c in contours:
                c = c[::simplify_stride]
                pts = np.stack([c[:, 0], c[:, 1], np.full(len(c), z)], axis=1).astype(np.float32)
                loops.append({"slice_index": int(z), "points": pts})

    elif axis == "y":
        idxs = np.linspace(margin, H - 1 - margin, n_slices).astype(int)
        for y in idxs:
            img = sdf[:, y, :]
            contours = measure.find_contours(img, level=level)
            for c in contours:
                c = c[::simplify_stride]
                pts = np.stack([c[:, 0], np.full(len(c), y), c[:, 1]], axis=1).astype(np.float32)
                loops.append({"slice_index": int(y), "points": pts})

    elif axis == "x":
        idxs = np.linspace(margin, D - 1 - margin, n_slices).astype(int)
        for x in idxs:
            img = sdf[x, :, :]
            contours = measure.find_contours(img, level=level)
            for c in contours:
                c = c[::simplify_stride]
                pts = np.stack([np.full(len(c), x), c[:, 0], c[:, 1]], axis=1).astype(np.float32)
                loops.append({"slice_index": int(x), "points": pts})

    else:
        raise ValueError("axis must be 'x', 'y', or 'z'")

    return loops
```

```python
class UNetLoopsDataset(Dataset):
    def __init__(
        self,
        npz_paths,
        target_shape=(128,128,128),
        axis="z",
        n_slices=30,
        loop_band=1.0,
        seed=0
    ):
        self.axis = axis
        self.n_slices = n_slices
        self.loop_band = loop_band
        self.rng = np.random.default_rng(seed)

        self.sdfs = []
        self.names = []
        for p in npz_paths:
            sdf = np.load(p)["sdf"].astype(np.float32)
            sdf = pad_crop_to_shape_3d(sdf, target_shape=target_shape)
            self.sdfs.append(sdf)
            self.names.append(os.path.basename(p))

    def __len__(self):
        return len(self.sdfs)

    def __getitem__(self, idx):
        sdf = self.sdfs[idx]
        name = self.names[idx]

        partial, mask, info = create_axis_aligned_loop_observation(
            sdf,
            axis=self.axis,
            n_slices=self.n_slices,
            loop_band=self.loop_band,
            margin=5
        )

        x = np.stack([partial, mask], axis=0).astype(np.float32)
        y = sdf[None].astype(np.float32)

        meta = {"name": name, "loop_info": info}
        return torch.from_numpy(x), torch.from_numpy(y), meta
```

```python
def build_graph_from_loops(loops, k_inter=3):
    X_list = []
    slice_id_list = []
    offsets = []
    cur = 0

    for L in loops:
        pts = L["points"]
        offsets.append(cur)
        cur += len(pts)
        X_list.append(pts)
        slice_id_list.append(np.full(len(pts), L["slice_index"], dtype=np.int32))

    if len(X_list) == 0:
        return np.zeros((0,3), np.float32), np.zeros((0,2), np.int64), np.zeros((0,), np.int32)

    X = np.concatenate(X_list, axis=0).astype(np.float32)
    slice_id = np.concatenate(slice_id_list, axis=0)

    edges = []

    # Intra-loop (cyclic)
    for off, L in zip(offsets, loops):
        n = len(L["points"])
        if n < 2:
            continue
        for i in range(n):
            a = off + i
            b = off + ((i + 1) % n)
            edges.append((a, b))
            edges.append((b, a))

    # Inter-slice (adjacent slices)
    unique_slices = sorted(set([L["slice_index"] for L in loops]))
    slice_to_nodes = {}

    node_cursor = 0
    for L in loops:
        n = len(L["points"])
        s = L["slice_index"]
        slice_to_nodes.setdefault(s, []).extend(range(node_cursor, node_cursor + n))
        node_cursor += n

    for s in unique_slices:
        for s2 in (s - 1, s + 1):
            if s2 not in slice_to_nodes:
                continue
            A = np.array(slice_to_nodes[s], dtype=np.int32)
            B = np.array(slice_to_nodes[s2], dtype=np.int32)
            if len(A) == 0 or len(B) == 0:
                continue

            nbrs = NearestNeighbors(n_neighbors=min(k_inter, len(B))).fit(X[B])
            _, idx = nbrs.kneighbors(X[A])

            for ai, neighs in enumerate(idx):
                a = int(A[ai])
                for nb in neighs:
                    b = int(B[int(nb)])
                    edges.append((a, b))
                    edges.append((b, a))

    E = np.array(edges, dtype=np.int64) if len(edges) > 0 else np.zeros((0,2), np.int64)
    return X, E, slice_id
```

```python
class GNNLoopsDataset(Dataset):
    def __init__(
        self,
        npz_paths,
        target_shape=(128,128,128),
        axis="z",
        n_slices=30,
        simplify_stride=2,
        k_inter=3,
        seed=0
    ):
        self.axis = axis
        self.n_slices = n_slices
        self.simplify_stride = simplify_stride
        self.k_inter = k_inter
        self.rng = np.random.default_rng(seed)

        self.sdfs = []
        self.names = []
        for p in npz_paths:
            sdf = np.load(p)["sdf"].astype(np.float32)
            sdf = pad_crop_to_shape_3d(sdf, target_shape=target_shape)
            self.sdfs.append(sdf)
            self.names.append(os.path.basename(p))

    def __len__(self):
        return len(self.sdfs)

    def __getitem__(self, idx):
        sdf = self.sdfs[idx]
        name = self.names[idx]

        loops = extract_loops_as_polylines(
            sdf,
            axis=self.axis,
            n_slices=self.n_slices,
            simplify_stride=self.simplify_stride,
            margin=5
        )

        X, E, slice_id = build_graph_from_loops(loops, k_inter=self.k_inter)

        X_t = torch.from_numpy(X)                      # (N,3)
        edge_index = torch.from_numpy(E).t().contiguous()  # (2,M) PyG-style
        slice_id_t = torch.from_numpy(slice_id)        # (N,)
        gt = torch.from_numpy(sdf[None].astype(np.float32))  # (1,D,H,W)

        meta = {"name": name, "n_loops": len(loops), "axis": self.axis}
        return X_t, edge_index, slice_id_t, gt, meta
```

```python
def gnn_collate_fn(batch):
    Xs, Es, sids, gts, metas = zip(*batch)
    return list(Xs), list(Es), list(sids), torch.stack(gts, dim=0), list(metas)
```

```python
def visualize_unet_sample_3d(dataset, idx=0, level=0.0, max_points=30000):
    """
    Visualize one UNetLoopsDataset sample:
      - GT surface (marching cubes)
      - Loop voxels (mask==1)

    Args:
      dataset: UNetLoopsDataset
      idx: sample index
      level: SDF iso-level (usually 0)
      max_points: subsample loop voxels for visualization
    """
    x, y, meta = dataset[idx]
    sdf = y[0].numpy()
    mask = x[1].numpy()

    # Extract GT surface
    verts, faces, _, _ = measure.marching_cubes(sdf, level=level)

    # Loop voxels
    pts = np.argwhere(mask > 0.5)
    if len(pts) > max_points:
        sel = np.random.choice(len(pts), size=max_points, replace=False)
        pts = pts[sel]

    fig = go.Figure()

    # GT surface
    fig.add_trace(go.Mesh3d(
        x=verts[:, 0], y=verts[:, 1], z=verts[:, 2],
        i=faces[:, 0], j=faces[:, 1], k=faces[:, 2],
        opacity=0.25,
        color="lightblue",
        name="GT surface"
    ))

    # Loop voxels
    fig.add_trace(go.Scatter3d(
        x=pts[:, 0], y=pts[:, 1], z=pts[:, 2],
        mode="markers",
        marker=dict(size=2, color="black"),
        name="Loop voxels"
    ))

    fig.update_layout(
        title=f"UNet sample | {meta['name']} | {meta['loop_info']}",
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=50, b=0)
    )
    fig.show()
```

```python
def visualize_gnn_sample_3d(dataset, idx=0, level=0.0, max_loops=200):
    """
    Visualize one GNNLoopsDataset sample:
      - GT surface
      - Loop polylines (closed curves)

    Args:
      dataset: GNNLoopsDataset
      idx: sample index
      level: SDF iso-level
      max_loops: max number of loops to display
    """
    X, edge_index, slice_id, gt, meta = dataset[idx]
    sdf = gt[0].numpy()

    # Extract GT surface
    verts, faces, _, _ = measure.marching_cubes(sdf, level=level)

    fig = go.Figure()

    # GT surface
    fig.add_trace(go.Mesh3d(
        x=verts[:, 0], y=verts[:, 1], z=verts[:, 2],
        i=faces[:, 0], j=faces[:, 1], k=faces[:, 2],
        opacity=0.30,
        color="lightblue",
        name="GT surface"
    ))

    # Group nodes by slice_id (each slice corresponds to one or more loops)
    unique_slices = np.unique(slice_id.numpy())
    shown = 0

    for s in unique_slices:
        if shown >= max_loops:
            break
        idxs = (slice_id.numpy() == s).nonzero()[0]
        pts = X[idxs].numpy()

        fig.add_trace(go.Scatter3d(
            x=pts[:, 0], y=pts[:, 1], z=pts[:, 2],
            mode="lines",
            line=dict(width=4),
            name=f"Loop @ slice {int(s)}" if shown == 0 else None,
            showlegend=(shown == 0)
        ))
        shown += 1

    fig.update_layout(
        title=f"GNN sample | {meta['name']} | loops={meta['n_loops']}",
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=50, b=0)
    )
    fig.show()
```

```python
class DoubleConv3D(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv3d(in_ch, out_ch, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv3d(out_ch, out_ch, 3, padding=1),
            nn.ReLU(inplace=True),
        )
    def forward(self, x):
        return self.net(x)

def match_size(src, ref):
    # Center crop or pad src so its (D,H,W) matches ref
    sd, sh, sw = src.shape[-3:]
    rd, rh, rw = ref.shape[-3:]

    d0 = max((sd - rd) // 2, 0)
    h0 = max((sh - rh) // 2, 0)
    w0 = max((sw - rw) // 2, 0)
    src = src[:, :, d0:d0+rd, h0:h0+rh, w0:w0+rw]

    sd, sh, sw = src.shape[-3:]
    pd = max(rd - sd, 0)
    ph = max(rh - sh, 0)
    pw = max(rw - sw, 0)
    if pd or ph or pw:
        src = F.pad(src, (pw//2, pw - pw//2, ph//2, ph - ph//2, pd//2, pd - pd//2))
    return src

class UNet3DSmallRobust(nn.Module):
    def __init__(self, in_ch=2, out_ch=1, base=16):
        super().__init__()
        self.down1 = DoubleConv3D(in_ch, base)
        self.pool1 = nn.MaxPool3d(2)
        self.down2 = DoubleConv3D(base, base*2)
        self.pool2 = nn.MaxPool3d(2)
        self.mid   = DoubleConv3D(base*2, base*4)

        self.up2   = nn.ConvTranspose3d(base*4, base*2, 2, stride=2)
        self.dec2  = DoubleConv3D(base*4, base*2)
        self.up1   = nn.ConvTranspose3d(base*2, base, 2, stride=2)
        self.dec1  = DoubleConv3D(base*2, base)

        self.out   = nn.Conv3d(base, out_ch, 1)

    def forward(self, x):
        x1 = self.down1(x)
        x2 = self.down2(self.pool1(x1))
        xm = self.mid(self.pool2(x2))

        x = self.up2(xm)
        x = match_size(x, x2)
        x = self.dec2(torch.cat([x, x2], dim=1))

        x = self.up1(x)
        x = match_size(x, x1)
        x = self.dec1(torch.cat([x, x1], dim=1))

        return self.out(x)
```

```python
def gradient_3d_forward(f):
    dx = f[:, :, 1:, :, :] - f[:, :, :-1, :, :]
    dy = f[:, :, :, 1:, :] - f[:, :, :, :-1, :]
    dz = f[:, :, :, :, 1:] - f[:, :, :, :, :-1]
    return dx, dy, dz

def eikonal_loss_3d(f):
    dx, dy, dz = gradient_3d_forward(f)
    dx_c = dx[:, :, :, :-1, :-1]
    dy_c = dy[:, :, :-1, :, :-1]
    dz_c = dz[:, :, :-1, :-1, :]
    grad_mag = torch.sqrt(dx_c**2 + dy_c**2 + dz_c**2 + 1e-6)
    return torch.mean(torch.abs(grad_mag - 1.0))

def laplacian_loss_3d(f):
    lap = (
        -6.0 * f
        + torch.roll(f,  1, dims=2) + torch.roll(f, -1, dims=2)
        + torch.roll(f,  1, dims=3) + torch.roll(f, -1, dims=3)
        + torch.roll(f,  1, dims=4) + torch.roll(f, -1, dims=4)
    )
    return torch.mean(lap**2)

def curvature_loss_3d(f):
    dxx = f[:, :, :, :, :-2] - 2*f[:, :, :, :, 1:-1] + f[:, :, :, :, 2:]
    dyy = f[:, :, :, :-2, :] - 2*f[:, :, :, 1:-1, :] + f[:, :, :, 2:, :]
    dzz = f[:, :, :-2, :, :] - 2*f[:, :, 1:-1, :, :] + f[:, :, 2:, :, :]
    return torch.mean(dxx**2) + torch.mean(dyy**2) + torch.mean(dzz**2)

def loss_function_3d(
    pred, gt, partial, mask,
    λ_rec=1.0,
    λ_mask=200.0,
    λ_eik=0.1,
    λ_smooth=0.02,
    λ_curv=0.02
):
    L_rec = torch.mean(torch.abs(pred - gt))
    L_mask = torch.mean(((pred - partial) ** 2) * mask)
    L_eik = eikonal_loss_3d(pred)
    L_smooth = laplacian_loss_3d(pred)
    L_curv = curvature_loss_3d(pred)

    return (
        λ_rec * L_rec
        + λ_mask * L_mask
        + λ_eik * L_eik
        + λ_smooth * L_smooth
        + λ_curv * L_curv
    )
```

```python

def train_unet(model, dataset, device="cuda", epochs=5, batch_size=1, lr=1e-3):
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=True)
    opt = torch.optim.Adam(model.parameters(), lr=lr)

    model.to(device)
    model.train()

    for ep in range(1, epochs+1):
        total = 0.0
        for x, gt, meta in loader:
            x = x.to(device)      # (B,2,D,H,W)
            gt = gt.to(device)    # (B,1,D,H,W)

            pred = model(x)

            partial = x[:, 0:1]
            mask    = x[:, 1:2]

            loss = loss_function_3d(pred, gt, partial, mask)

            opt.zero_grad()
            loss.backward()
            opt.step()

            total += float(loss.item())

        print(f"Epoch {ep}/{epochs} | loss={total/len(loader):.4f}")
```

```python
def safe_marching_cubes(vol, preferred_level=0.0):
    vmin, vmax = float(vol.min()), float(vol.max())
    if vmin <= preferred_level <= vmax:
        return None
    # else:
    #     level = 0.5 * (vmin + vmax)
    verts, faces, _, _ = measure.marching_cubes(vol, level=preferred_level)
    return verts, faces, preferred_level, (vmin, vmax)

def visualize_prediction_vs_gt_safe(model, dataset, idx=0, device="cpu", shift=160):
    model.eval()
    with torch.no_grad():
        x, gt, meta = dataset[idx]
        pred = model(x.unsqueeze(0).to(device)).cpu()[0,0].numpy()
        gt_np = gt[0].numpy()

    v_gt, f_gt, lvl_gt, rg_gt = safe_marching_cubes(gt_np, preferred_level=0.0)
    v_pr, f_pr, lvl_pr, rg_pr = safe_marching_cubes(pred, preferred_level=0.0)

    v_pr_shifted = v_pr.copy()
    v_pr_shifted[:, 0] += shift

    fig = go.Figure()
    fig.add_trace(go.Mesh3d(
        x=v_gt[:,0], y=v_gt[:,1], z=v_gt[:,2],
        i=f_gt[:,0], j=f_gt[:,1], k=f_gt[:,2],
        opacity=0.5, name=f"GT (level={lvl_gt:.3f})"
    ))
    fig.add_trace(go.Mesh3d(
        x=v_pr_shifted[:,0], y=v_pr_shifted[:,1], z=v_pr_shifted[:,2],
        i=f_pr[:,0], j=f_pr[:,1], k=f_pr[:,2],
        opacity=0.5, name=f"Pred (level={lvl_pr:.3f})"
    ))

    fig.update_layout(
        title=(f"{meta['name']} | {meta['loop_info']}<br>"
               f"GT range={rg_gt} | Pred range={rg_pr} | Pred level used={lvl_pr:.3f}"),
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=60, b=0)
    )
    fig.show()
```

```python
import torch
from torch_geometric.data import Data
from torch_geometric.utils import negative_sampling

def sample_to_pyg_data(X, edge_index):
    """
    X: (N,3) tensor
    edge_index: (2,M) tensor
    """
    return Data(x=X, edge_index=edge_index)

def make_link_prediction_batch(data: Data, num_neg=None):
    """
    Returns:
      pos_edge_index, neg_edge_index
    """
    pos_edge_index = data.edge_index
    if num_neg is None:
        num_neg = pos_edge_index.size(1)

    neg_edge_index = negative_sampling(
        edge_index=pos_edge_index,
        num_nodes=data.num_nodes,
        num_neg_samples=num_neg
    )
    return pos_edge_index, neg_edge_index
```

```python
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv

class GraphSAGEEncoder(nn.Module):
    def __init__(self, in_dim=3, hidden=64, out_dim=64):
        super().__init__()
        self.conv1 = SAGEConv(in_dim, hidden)
        self.conv2 = SAGEConv(hidden, out_dim)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return x

class DotProductDecoder(nn.Module):
    def forward(self, z, edge_index):
        # z: (N, F), edge_index: (2, E)
        src, dst = edge_index
        return (z[src] * z[dst]).sum(dim=1)  # logits

class LinkPredModel(nn.Module):
    def __init__(self, in_dim=3, hidden=64, out_dim=64):
        super().__init__()
        self.encoder = GraphSAGEEncoder(in_dim, hidden, out_dim)
        self.decoder = DotProductDecoder()

    def forward(self, data, pos_edge_index, neg_edge_index):
        z = self.encoder(data.x, data.edge_index)
        pos_logits = self.decoder(z, pos_edge_index)
        neg_logits = self.decoder(z, neg_edge_index)
        return pos_logits, neg_logits
```

```python
import torch

def link_pred_loss(pos_logits, neg_logits):
    pos_loss = F.binary_cross_entropy_with_logits(pos_logits, torch.ones_like(pos_logits))
    neg_loss = F.binary_cross_entropy_with_logits(neg_logits, torch.zeros_like(neg_logits))
    return pos_loss + neg_loss

def train_gnn_linkpred(model, gnn_dataset, device="cuda", epochs=10, lr=1e-3):
    model.to(device)
    opt = torch.optim.Adam(model.parameters(), lr=lr)

    for ep in range(1, epochs+1):
        model.train()
        total = 0.0

        for i in range(len(gnn_dataset)):
            X, edge_index, slice_id, gt, meta = gnn_dataset[i]
            if X.shape[0] == 0 or edge_index.shape[1] == 0:
                continue

            data = sample_to_pyg_data(X, edge_index).to(device)
            pos_edge_index, neg_edge_index = make_link_prediction_batch(data)

            pos_logits, neg_logits = model(data, pos_edge_index, neg_edge_index)
            loss = link_pred_loss(pos_logits, neg_logits)

            opt.zero_grad()
            loss.backward()
            opt.step()

            total += float(loss.item())

        denom = max(1, len(gnn_dataset))
        print(f"Epoch {ep}/{epochs} | loss={total/denom:.4f}")
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
gnn_model = LinkPredModel(in_dim=3, hidden=64, out_dim=64)

train_gnn_linkpred(gnn_model, gnn_dataset, device=device, epochs=60, lr=1e-3)
```

```python
import numpy as np

@torch.no_grad()
def predict_topk_links_between_adjacent_slices(model, gnn_dataset, idx=0, k=50, device="cpu"):
    model.eval()
    X, edge_index, slice_id, gt, meta = gnn_dataset[idx]
    if X.shape[0] == 0:
        print("No nodes in this sample.")
        return None

    data = sample_to_pyg_data(X, edge_index).to(device)
    z = model.encoder(data.x, data.edge_index)

    sid = slice_id.numpy()
    unique_slices = np.unique(sid)
    if len(unique_slices) < 2:
        print("Not enough slices.")
        return None

    # pick two adjacent slices (by index)
    s0 = unique_slices[len(unique_slices)//2]
    s1 = s0 + 1
    if s1 not in unique_slices:
        s1 = unique_slices[max(0, len(unique_slices)//2 - 1)]

    A = np.where(sid == s0)[0]
    B = np.where(sid == s1)[0]
    if len(A) == 0 or len(B) == 0:
        print("Empty slice sets.")
        return None

    # compute scores for all pairs (A x B) (can be heavy; keep slices modest)
    ZA = z[A]
    ZB = z[B]
    scores = ZA @ ZB.t()  # (|A|,|B|)
    flat = scores.flatten()
    topk = torch.topk(flat, k=min(k, flat.numel()))

    pairs = []
    for val, idx_flat in zip(topk.values.cpu().numpy(), topk.indices.cpu().numpy()):
        ia = idx_flat // len(B)
        ib = idx_flat % len(B)
        pairs.append((int(A[ia]), int(B[ib]), float(val)))

    print(f"Top-{len(pairs)} predicted links between slices {int(s0)} and {int(s1)}")
    return pairs
```
